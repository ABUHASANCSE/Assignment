### Exercise 9.1: Safe Feature Division (`ZeroDivisionError`)
In feature engineering, dividing two numeric features can crash if the denominator is zero.
Write a function `safe_feature_ratio(numerator, denominator)`:
- In `try`: returns `round(numerator / denominator, 2)`
- In `except ZeroDivisionError`: prints `"Warning: Denominator is 0. Returning 0.0"` and returns `0.0`
- Test with `safe_feature_ratio(150, 3)` and `safe_feature_ratio(150, 0)`.

**Expected Output:**
```text
Ratio 1: 50.0
Warning: Denominator is 0. Returning 0.0
Ratio 2: 0.0
```


````python
# TODO: Define safe_feature_ratio(numerator, denominator) with try-except
def safe_feature_ratio(numerator, denominator):
    try:
        return round(numerator / denominator, 2)
    except ZeroDivisionError:
        print("Warning: Denominator is 0. Returning 0.0")
        return 0.0


ratio = safe_feature_ratio(150, 3)
print(f"Ratio 1: {ratio}")

ratio2 = safe_feature_ratio(150, 0)
print(f"Ratio 2: {ratio2}")
````
