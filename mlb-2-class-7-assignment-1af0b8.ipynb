# Outliers

# 1. find and mitigate(trimming and capping both) outliers on age column of diabetes dataset

```python
diabetes.head()
```

```python
diabetes.isnull().sum()
```

```python
diabetes['age'].describe()
```

```python
Q1 = diabetes['age'].quantile(0.25)
Q3 = diabetes['age'].quantile(0.75)
IQR = Q3 - Q1
print(Q1)
print(Q3)
print(IQR)

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
outliers =  diabetes[( diabetes['age'] < lower_bound) |( diabetes['age'] > upper_bound)]
print(outliers)
#NO outlier
```

# 2. find and mitigate(trimming and capping both) outliers on price column of house dataset

```python
house.head()
```

```python
house.isnull().sum()
```

```python
house['price'].describe()
```

```python
house_Q1=house['price'].quantile(0.25)
house_Q3 = house['price'].quantile(0.75)
house_IQR = house_Q3 - house_Q1
print(house_Q1)
print(house_Q3)
print(house_IQR)

house_lower_bound = house_Q1- 1.5 * house_IQR
house_upper_bound =house_Q3 + 1.5 * house_IQR

print("Lower bound:", house_lower_bound)
print("Upper bound:", house_upper_bound)
```

```python
house_outliers =  house[( house['price'] < house_lower_bound) |( house['price'] > house_upper_bound)]
print(house_outliers)

print("outlier :",house_outliers.shape)
#outlier 240 
```

```python
#trimmed house price column
house_price_trimmed = house[
    (house['price'] >= house_lower_bound) &
    (house['price'] <= house_upper_bound)
]
house_price_trimmed
```

```python
#capping 
house_capped = house.copy()

house_capped['price'] = house_capped['price'].clip(lower=house_lower_bound,upper=house_upper_bound)
```

```python
capped_outliers = house_capped[
    (house_capped['price'] < house_lower_bound) |
    (house_capped['price'] > house_upper_bound)
]

len(capped_outliers)
```
