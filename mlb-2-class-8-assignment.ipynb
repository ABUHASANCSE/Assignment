```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('/kaggle/input/datasets/lovishbansal123/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
```

# Univariate Analysis

# 1. Pie chart

```python
titanic['Pclass'].value_counts().plot.pie(autopct='%1.1f%%')
plt.title('Passenger Class Distribution')
plt.ylabel('')
plt.show()
```

# 2. Count plot

```python
sns.countplot(x='Sex', data=titanic)
plt.show()
```

# 3. Kde + Rug plot

```python
sns.kdeplot(titanic['Age'], fill=True)
sns.rugplot(titanic['Age'])

plt.show()
```

# 4. Violin plot

```python
sns.violinplot(y='Fare', data=titanic)
plt.show()
```

# 5. Box plot

```python
sns.boxplot(y='total_bill', data=tips)
plt.show()
```

# 6. Boxen plot

```python
sns.boxenplot(y='tip', data=tips)
plt.show()
```

# 7. Histogram

```python
plt.hist(titanic['Age'].dropna())
plt.show()
```

# Bivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(x='total_bill', y='tip', data=tips)

plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Total Bill vs Tip')
plt.show()
```

# 2. Line plot

```python
sns.lineplot(x='total_bill', y='tip', data=tips)

plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Total Bill vs Tip')
plt.show()
```

# 3. Joint plot

```python
sns.jointplot(x='total_bill', y='tip', data=tips)
plt.show()
```

# 4. Hexbin plot

```python
plt.hexbin(tips['total_bill'], tips['tip'], gridsize=20)

plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Total Bill vs Tip')
plt.colorbar(label='Count')
plt.show()
```

# 5. 2D kde plot

```python
sns.kdeplot(
    x='sepal_length',
    y='sepal_width',
    data=iris,
    hue='species',
    fill=True
)

plt.show()
```

# 6. Box plot

```python
sns.boxplot(x='sex', y='total_bill', data=tips)
plt.show()
```

# 7. Violin plot

```python
sns.violinplot(x='sex', y='tip', data=tips)
plt.show()
```

# 8. Swarm plot

```python
sns.swarmplot(x='Sex', y='Age', data=titanic)
plt.show()
```

# 9. Count plot

```python
sns.countplot(x='Sex', hue='Survived', data=titanic)
plt.show()
```

# Multivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    hue='species',
    size='petal_length',
    data=iris
)

plt.show()
```

# 2. Pair plot

```python
sns.pairplot(iris, hue='species')
plt.show()
```

# 3. Heatmap

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    titanic.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)

plt.title('Titanic Correlation Heatmap')
plt.show()
```
